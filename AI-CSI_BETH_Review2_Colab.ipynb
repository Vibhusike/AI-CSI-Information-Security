```python
import json, pandas as pd
from pathlib import Path
results=json.loads(Path("results/beth_review2_results.json").read_text())
rows=[]
for name, item in results["models"].items():
    rows.append({"model":name,"threshold":item["threshold_from_validation"],**item["test"]})
display(pd.DataFrame(rows)[["model","accuracy","precision_evil","recall_evil","f1_evil","macro_f1"]])
for name,item in results["models"].items():
    print(name, item["test"]["confusion_matrix_normal_evil"])
```
